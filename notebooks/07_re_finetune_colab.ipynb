## 3-1. BatteryDataset

```python
import numpy as np
import pandas as pd
from PIL import Image
import torch
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms.functional as TF
from torchvision import transforms as T

IMG_SIZE = 513
NUM_CLASSES = 3   # bg, Pollution, Damaged

class BatteryDataset(Dataset):
    def __init__(self, split_csv, img_dir, mask_dir, augment=False):
        self.names = pd.read_csv(split_csv, header=None)[0].tolist()
        self.img_dir = Path(img_dir)
        self.mask_dir = Path(mask_dir)
        self.augment = augment
        self.normalize = T.Normalize(mean=(0.485, 0.456, 0.406),
                                     std=(0.229, 0.224, 0.225))

    def __len__(self):
        return len(self.names)

    def __getitem__(self, idx):
        name = self.names[idx]
        img = Image.open(next(self.img_dir.glob(f'{name}.*'))).convert('RGB')
        mask = Image.open(self.mask_dir / f'{name}.png')

        img  = TF.resize(img,  [IMG_SIZE, IMG_SIZE])
        mask = TF.resize(mask, [IMG_SIZE, IMG_SIZE],
                         interpolation=TF.InterpolationMode.NEAREST)

        if self.augment:
            if torch.rand(1) < 0.5:
                img, mask = TF.hflip(img), TF.hflip(mask)
            if torch.rand(1) < 0.3:
                img = T.ColorJitter(brightness=0.2, contrast=0.2)(img)

        img  = TF.to_tensor(img)
        img  = self.normalize(img)
        mask = torch.as_tensor(np.array(mask), dtype=torch.long)
        return img, mask


train_ds = BatteryDataset(SPLIT_DIR / 'train.csv', IMG_DIR, MASK_DIR, augment=True)
val_ds   = BatteryDataset(SPLIT_DIR / 'val.csv',   IMG_DIR, MASK_DIR, augment=False)

train_loader = DataLoader(train_ds, batch_size=8, shuffle=True,  num_workers=2, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=8, shuffle=False, num_workers=2, pin_memory=True)

print(f'Train: {len(train_ds)}장,  Val: {len(val_ds)}장')
```

## 4. 모델 로드 + 헤드 교체 + freeze

AI허브 가중치(4채널)를 로드한 뒤 **마지막 1×1 conv만 3채널로 재초기화**하고,<br/> backbone+ASPP의 나머지는 **freeze**. 새 헤드 + decoder만 학습합니다.

## 4-1. 헤드 교체

```python
import torch.nn as nn

# 1. 모델 인스턴스 (NUM_CLASSES=4 로 일단 생성 → 가중치 그대로 로드 가능)
model = DeepLab(num_classes=4, backbone='drn', output_stride=16,
                sync_bn=False, freeze_bn=True)

ckpt = torch.load(str(PT_PATH), map_location='cpu', weights_only=False)
result = model.load_state_dict(ckpt['state_dict'], strict=False)
print(f'AI허브 가중치 로드 — Missing: {len(result.missing_keys)}, Unexpected: {len(result.unexpected_keys)}')

# 2. 마지막 1x1 conv (decoder.last_conv.8) 를 3채널로 교체
old = model.decoder.last_conv[8]
new = nn.Conv2d(old.in_channels, NUM_CLASSES,
                kernel_size=old.kernel_size, stride=old.stride,
                padding=old.padding, bias=True)
nn.init.kaiming_normal_(new.weight, mode='fan_out', nonlinearity='relu')
nn.init.zeros_(new.bias)
model.decoder.last_conv[8] = new
print(f'헤드 교체: out_channels {old.out_channels} → {new.out_channels}')

# 3. backbone + ASPP freeze (decoder만 학습)
for p in model.backbone.parameters(): p.requires_grad = False
for p in model.aspp.parameters():     p.requires_grad = False

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total     = sum(p.numel() for p in model.parameters())
print(f'학습 파라미터: {trainable:,} / 전체 {total:,} ({100*trainable/total:.1f}%)')

model = model.cuda()
```

> **이유**:
> - Backbone(DRN-D-54)은 ImageNet에서 학습된 풍부한 시각 표현 → 작은 데이터(249장)에서 더 학습시키면 과적합 위험.
> - ASPP도 multi-scale feature 추출 능력이 일반적이라 freeze 유지.
> - Decoder + 새 헤드만 학습 → **클래스 의미 재매핑**에 집중.
> - 1단계로 부족하면 ASPP까지 풀고 재학습 (셀 분기 시나리오 참조).

## 5. 손실 함수 & Optimizer

EDA에서 확인된 클래스 불균형(Pollution 80% / Damaged 20%)에 대응하여 **class weight** 적용합니다.<br/>
학습률은 decoder와 새 헤드를 분리해 헤드를 더 빠르게 학습합니다.

### 5-1. 실제 픽셀 빈도로 class weight 산정

```python
counts = np.zeros(NUM_CLASSES, dtype=np.int64)
for name in train_ds.names:
    m = np.array(Image.open(MASK_DIR / f'{name}.png'))
    for c in range(NUM_CLASSES):
        counts[c] += int((m == c).sum())

freq = counts / counts.sum()
raw  = 1.0 / np.log(1.02 + freq)
# [v2] v1은 결함:배경 가중치가 ~30배 → FP penalty가 거의 없어 과예측 유발.
#      결함 가중치를 배경의 최대 5배로 상한 → 미검 88%. 과교정으로 확인
#      배경 가중치 상한 20배 진행
MAX_RATIO = 20.0
raw = np.minimum(raw, raw[0] * MAX_RATIO)
weights = raw / raw.sum() * NUM_CLASSES

print(f'클래스 픽셀 비율 : {freq}')
print(f'적용 weight      : {weights}')
print(f'결함:배경 비율   : Pollution {weights[1]/weights[0]:.1f}배, '
    f'Damaged {weights[2]/weights[0]:.1f}배  (v1: 29·33배)')

class_weights = torch.tensor(weights, dtype=torch.float32).cuda()
```

### 5-2. Loss / Optimizer / Scheduler

```python
# import torch.nn.functional as F

# [v2 추가] 다중 클래스 soft Dice loss.
# 과예측 시 예측 영역(card)이 커져 Dice score가 떨어짐 → FP에 직접 penalty.
# class DiceLoss(nn.Module):
#     def __init__(self, num_classes, smooth=1.0, ignore_bg=True):
#         super().__init__()
#         self.num_classes = num_classes
#         self.smooth = smooth
#         self.ignore_bg = ignore_bg

#     def forward(self, logits, target):
#         probs  = torch.softmax(logits, dim=1)                        # (N,C,H,W)
#         onehot = F.one_hot(target, self.num_classes).permute(0, 3, 1, 2).float()
#         dims   = (0, 2, 3)
#         inter  = (probs * onehot).sum(dims)
#         card   = probs.sum(dims) + onehot.sum(dims)
#         dice   = (2 * inter + self.smooth) / (card + self.smooth)    # (C,)
#         if self.ignore_bg:
#             dice = dice[1:]                                          # 결함 클래스만
#         return 1.0 - dice.mean()

# ce_loss   = nn.CrossEntropyLoss(weight=class_weights)
# dice_loss = DiceLoss(NUM_CLASSES, ignore_bg=True)

# [v2] v1: CrossEntropyLoss 단독 → CE(약화된 가중치) + Dice(과예측 억제)
# def criterion(logits, target):
#     return ce_loss(logits, target) + dice_loss(logits, target)

# [재시도] Dice 제거 — 단일 노브(class weight)로만 튜닝.
#          v1과 동일한 CE-only 구조, 가중치 상한만 다름(20배).
criterion = nn.CrossEntropyLoss(weight=class_weights)

new_head_params = list(model.decoder.last_conv[8].parameters())
decoder_other = [p for n, p in model.decoder.named_parameters()
                 if not n.startswith('last_conv.8') and p.requires_grad]

optimizer = torch.optim.AdamW([
    {'params': new_head_params, 'lr': 1e-3},
    {'params': decoder_other,   'lr': 1e-4},
], weight_decay=1e-4)

EPOCHS = 30
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
```

### 5-3. class weight 튜닝 경과

  클래스 불균형(배경 99.4% / 결함 0.6%)에 대응하는 class weight 설정이 모델의
  과검·미검 경향을 좌우합니다. 검증 결과를 반영하여 아래와 같이 단계적으로 조정하였습니다.

  | 단계 | 결함:배경 가중치 | Loss | 결과 (test 결함 14장) | 판정 |
  |---|---|---|---|---|
  | v1 | 약 30배 | CE | Recall 0.779 / Precision 0.327 / **Inst. Recall 0.696** | 과검 |
  | 1차 보정 | 5배 상한 + Dice | CE+Dice | Recall 0.177 / Precision 0.777 / **Inst. Recall 0.116** | 과교정(미검) |
  | **2차 보정** | **20배 상한** | **CE** | Recall 0.623 / Precision 0.432 / **Inst. Recall 0.503** | 균형 확보 - 검증 시도 예정 |

  - **v1**은 결함:배경 가중치가 약 30배로, loss가 미검(FN)을 과검(FP)보다 훨씬 강하게 처벌하여 모델이 "애매하면 결함"으로 과예측하였습니다.
  - **1차 보정**에서 가중치 상한을 5배로 낮추고 Dice Loss를 추가하였으나, 두 변경이 모두 "배경 예측" 방향으로 작용하여 **과교정**이 발생했습니다.
  
  과검은 줄었지만 실제 결함검출률(Instance Recall)이 0.696 → 0.116으로 붕괴하여, 결함의 약 88%를 놓쳤습니다.
  
  검사 공정에서 미검은 과검보다 치명적이므로 채택할 수 없는 결과입니다.

  - **2차 보정**에서는 변수를 하나로 좁혀, Dice Loss를 제거하고 v1과 동일한 CE 단독 구조로 되돌린 뒤 **가중치 상한만 20배**로 설정하였습니다.
  
  v1(30배)과 1차 보정(5배)의 중간값으로, **결함 검출률(recall) 보호를 우선**하면서 과검을 점진적으로 줄이는 것이 목표입니다.

## 6. 학습 루프

epoch마다 train loss / val loss / val mIoU 출력합니다.<br/>
val mIoU가 가장 높은 가중치를 best로 저장합니다.<br/>
AMP 적용으로 T4에서 약 1.7배 가속합니다.

### 6-1. 학습 루프

```python
from torch.amp import autocast, GradScaler

scaler = GradScaler('cuda')

def compute_miou(pred, target, num_classes):
    ious = []
    for c in range(num_classes):
        p = (pred == c); t = (target == c)
        inter = (p & t).sum().item()
        union = (p | t).sum().item()
        if union > 0:
            ious.append(inter / union)
    return float(np.mean(ious)) if ious else 0.0

best_miou = -1.0
history = []

for epoch in range(EPOCHS):
    model.train()
    for m in model.modules():
        if isinstance(m, nn.BatchNorm2d):
            m.eval()                       # freeze_bn=True 유지

    tr_loss = 0.0
    for img, mask in train_loader:
        img, mask = img.cuda(non_blocking=True), mask.cuda(non_blocking=True)
        optimizer.zero_grad()
        with autocast('cuda'):
            logits = model(img)
            loss = criterion(logits, mask)
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        tr_loss += loss.item() * img.size(0)
    tr_loss /= len(train_ds)

    model.eval()
    val_loss, val_miou = 0.0, 0.0
    val_clean, val_overkill = 0, 0          # [v2] 정상품 과검 추적
    with torch.no_grad():
        for img, mask in val_loader:
            img, mask = img.cuda(), mask.cuda()
            with autocast('cuda'):
                logits = model(img)
                loss = criterion(logits, mask)
            val_loss += loss.item() * img.size(0)
            pred = logits.argmax(dim=1)
            val_miou += compute_miou(pred, mask, NUM_CLASSES) * img.size(0)
             # [v2] 정상품(GT 전부 배경) 중 결함으로 예측된 장수 = 과검
            for b in range(img.size(0)):
                if (mask[b] > 0).sum() == 0:
                    val_clean += 1
                    if (pred[b] > 0).sum() > 0:
                        val_overkill += 1

    val_loss /= len(val_ds)
    val_miou /= len(val_ds)
    overkill_rate = val_overkill / max(val_clean, 1)     # [v2]
    scheduler.step()

    history.append({'epoch': epoch, 'tr_loss': tr_loss, 'val_loss': val_loss, 'val_miou': val_miou, 'overkill_rate': overkill_rate})
    print(f'[{epoch+1:02d}/{EPOCHS}] tr_loss={tr_loss:.4f}  val_loss={val_loss:.4f}  val_mIoU={val_miou:.4f}')

    if val_miou > best_miou:
        best_miou = val_miou
        torch.save({'state_dict': model.state_dict(),
                    'epoch': epoch, 'val_miou': val_miou,
                    'overkill_rate': overkill_rate,
                    'num_classes': NUM_CLASSES},
                   str(OUT_DIR / 'best_finetuned_v2.pt'))
        print(f'  → best 저장 (mIoU={best_miou:.4f}), 과검율={overkill_rate*100:.1f}%)')

print(f'\n학습 완료. best val mIoU = {best_miou:.4f}')
```

 - **2차 보정**에서는 변수를 하나로 좁혀, Dice Loss를 제거하고 v1과 동일한 CE 단독 구조로 되돌린 뒤 가중치 상한만 20배로 설정하였습니다.<br/>
 그 결과 Pixel Recall 0.623 / Precision 0.432 / Instance Recall 0.503으로, v1(과검)과 1차 보정(과교정) 사이의 균형 지점을 확보하였습니다.<br/>
 다만 Instance Recall 0.503은 결함 인스턴스의 절반가량을 놓치는 한계이며, 후처리를 포함한 최종 검증으로 채택 여부를 판정할 예정입니다.

## 8. Fine-tune 전후 정량 비교

### 8-1. Fine-tune 모델 평가

```python
from skimage.measure import label, regionprops
import torch.nn.functional as F

best = torch.load(str(OUT_DIR / 'best_finetuned_v2.pt'), map_location='cuda', weights_only=False)
model.load_state_dict(best['state_dict'])
model.eval()

test_names = pd.read_csv(SPLIT_DIR / 'test.csv', header=None)[0].tolist()
test_defect = [n for n in test_names
               if (np.array(Image.open(MASK_DIR / f'{n}.png')) > 0).any()]
print(f'평가 결함 샘플(test split): {len(test_defect)}장')

normalize = T.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225))
pix_tp = pix_fn = pix_fp = 0
inst_total = inst_hit = 0

for name in test_defect:
    img = Image.open(next(IMG_DIR.glob(f'{name}.*'))).convert('RGB')
    gt = (np.array(Image.open(MASK_DIR / f'{name}.png')) > 0).astype(np.uint8)

    x = normalize(TF.to_tensor(TF.resize(img, [IMG_SIZE, IMG_SIZE]))).unsqueeze(0).cuda()
    with torch.no_grad():
        logits = model(x)
    logits_up = F.interpolate(logits, size=gt.shape, mode='bilinear', align_corners=False)
    pred = (logits_up.argmax(dim=1)[0].cpu().numpy() > 0).astype(np.uint8)

    pix_tp += int(((pred == 1) & (gt == 1)).sum())
    pix_fn += int(((pred == 0) & (gt == 1)).sum())
    pix_fp += int(((pred == 1) & (gt == 0)).sum())
    lab = label(gt)
    for region in regionprops(lab):
        inst_total += 1
        if pred[region.coords[:, 0], region.coords[:, 1]].any():
            inst_hit += 1

p_rec = pix_tp / max(pix_tp + pix_fn, 1)
p_pre = pix_tp / max(pix_tp + pix_fp, 1)
i_rec = inst_hit / max(inst_total, 1)

print(f'\n=== Fine-tune 후 성능 (test split) ===')
print(f'  Pixel  Recall   : {p_rec:.3f}')
print(f'  Pixel  Precision: {p_pre:.3f}')
print(f'  Instance Recall : {i_rec:.3f}')

before = {'p_rec': 0.312, 'p_pre': 0.048, 'i_rec': 0.696}
print(f'\n=== Before / After ===')
print(f'              Before   After')
print(f'  Pix Recall  {before["p_rec"]:.3f}    {p_rec:.3f}')
print(f'  Pix Precis  {before["p_pre"]:.3f}    {p_pre:.3f}')
print(f'  Inst Recall {before["i_rec"]:.3f}    {i_rec:.3f}')
```

### 학습 개요

AI허브에서 제공된 DeepLabV3+ 사전학습 가중치는 4클래스(`bg / cell / Pollution / Damaged` 추정) 체계로
학습되어, 본 프로젝트의 라벨 체계(3클래스)와 클래스 의미가 불일치합니다. 03번 노트북의 정량 분석에서 **셀
본체가 결함 confidence에 포함되는 object prior leakage**가 확인되어, 마지막 1×1 conv 헤드만 교체한 뒤
fine-tune을 수행하였습니다.

| 항목 | 설정 |
|---|---|
| 환경 | Google Colab Pro, Tesla T4 (14.6 GB VRAM) |
| 학습 파라미터 | 1,304,419 / 전체 40,732,435 (3.2%) |

### class weight 튜닝 경과 요약

초기 학습은 결함:배경 가중치가 약 30배로 설정되어 loss가 미검을 과검보다 과도하게 처벌한 결과, 정상 영역까지
결함으로 예측하는 **과검** 문제가 발생하였습니다. 이를 완화하기 위한 1차 보정(가중치 5배 + Dice Loss)은
반대로 **과교정**되어 결함의 약 88%를 놓쳤습니다. 본 학습은 변수를 class weight 하나로 좁히고 상한을 20배로
설정한 **2차 보정** 결과입니다. (상세 경과는 5-3 절 참조)

### 정량 결과 (test split, 결함 보유 14장)

| 지표 | Before (raw AI허브) | After (2차 보정) | 개선폭 |
|---|---|---|---|
| Pixel Recall | 0.312 | **0.623** | +0.311 (약 2배) |
| Pixel Precision | 0.048 | **0.432** | +0.384 (약 9배) |
| Instance Recall | 0.696 | 0.503 | −0.193 |

### 결과 해석

- **Pixel Precision이 0.048에서 0.432로 약 9배 개선**되어, 사전학습 모델의 핵심 문제였던 object prior
leakage(셀 본체 전체가 결함으로 표시되던 현상)가 크게 완화되었습니다.
- **Pixel Recall이 0.312에서 0.623으로 상승**하여, 결함 영역의 약 62%를 픽셀 단위로 검출합니다.
- **Instance Recall은 0.696에서 0.503으로 하락**하였습니다. raw 사전학습 모델의 높은 Instance Recall은 결함
여부와 무관하게 광범위하게 과예측한 결과 우연히 결함 영역에 걸친 것으로, 실제 검출 능력으로 보기는
어렵습니다. 본 모델은 보다 선별적으로 예측하며 결함 인스턴스의 약 절반을 명확히 검출합니다. 다만 나머지
절반을 놓치는 점은 본 모델의 한계이며, 미검은 후처리로 복구할 수 없으므로 검증 단계에서 별도로 다룹니다.

### 학습 안정성

- val_loss가 epoch 30까지 단조 감소(0.769 → 0.134)하여 발산·심각한 과적합 징후는 없습니다.
- val_mIoU는 epoch 11에서 0.4005로 최고치에 도달한 뒤 후반부에 0.398 수준으로 소폭 하락하였으며, best
가중치는 mIoU 최고 시점(epoch 11)을 저장하였습니다.
- Backbone·ASPP 동결로 학습 파라미터를 3.2%로 제한하여, 174장 소규모 데이터셋에서도 안정적으로
학습되었습니다.

### 결론 및 다음 단계

본 2차 보정 모델은 Pixel Precision 0.432로 사전 정의한 기준(≥ 0.30)을 충족하였으나, Pixel Recall은 0.623으로
기준(≥ 0.70)에 다소 미치지 못하였습니다. 이는 과검과 미검의 균형을 맞추는 과정에서 검출률을 일부 양보한
결과입니다. 또한 raw 출력 기준 Instance Recall 0.503으로 결함 인스턴스의 약 절반을 놓치는 한계가
확인되었으므로, 본 가중치의 최종 채택 여부는 다음 단계에서 결정합니다.

- **ONNX 변환** (05번 노트북): 본 가중치를 `battery_deeplab_v2.onnx`로 변환
- **검증** (06번 노트북): 후처리(confidence threshold, morphology)를 포함한 실제 추론 파이프라인에서 정상품
과검율과 결함 검출률을 측정하고, 기존 v1 모델과 정량 비교
- 검증 결과에 따라 본 모델 채택, 또는 추가 개선(ASPP 동결 해제 등 구조 변경) 여부를 판정하겠습니다.
